```python
target_text = (
    df["Credit Risk"]
    .astype(str)
    .str.strip()
    .str.lower()
)

target_mapping = {
    "1": 0,       # Good risk in original UCI coding
    "2": 1,       # Bad risk in original UCI coding
    "good": 0,
    "bad": 1
}

y = target_text.map(target_mapping)

if y.isna().any():
    raise ValueError(
        "Unexpected target labels. Check the original target values."
    )

X = df.drop(columns=["Credit Risk"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

numeric_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight=None
    ))
])

print("Preprocessing and model pipeline created.")
```

```python
model.fit(X_train, y_train)

print("Model training complete!")
```

```python
y_pred = model.predict(X_test)

print("Actual labels:")
print(y_test.head(10).tolist())

print("\nPredicted labels:")
print(y_pred[:10].tolist())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred, zero_division=0
)

recall = recall_score(
    y_test, y_pred, zero_division=0
)

f1 = f1_score(
    y_test, y_pred, zero_division=0
)

print(f"Accuracy:  {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall:    {recall * 100:.2f}%")
print(f"F1-score:  {f1 * 100:.2f}%")

print("\nFull classification report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Good credit risk", "Bad credit risk"],
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Good risk", "Bad risk"],
    yticklabels=["Good risk", "Bad risk"]
)

plt.xlabel("Predicted label")
plt.ylabel("Actual label")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Get the preprocessing pipeline and trained Random Forest
preprocessor = model.named_steps["preprocessing"]
rf_model = model.named_steps["classifier"]

# Convert categorical columns into numerical features
X_train_transformed = preprocessor.transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

# Convert sparse matrices into normal arrays if needed
if hasattr(X_train_transformed, "toarray"):
    X_train_transformed = X_train_transformed.toarray()

if hasattr(X_test_transformed, "toarray"):
    X_test_transformed = X_test_transformed.toarray()

# Get the names of the transformed features
feature_names = preprocessor.get_feature_names_out()

# Make readable tables for SHAP
X_train_transformed = pd.DataFrame(
    X_train_transformed,
    columns=feature_names
)

X_test_transformed = pd.DataFrame(
    X_test_transformed,
    columns=feature_names
)

# Select a small background sample
background = X_train_transformed.sample(
    n=min(50, len(X_train_transformed)),
    random_state=42
)

# Explain a few test examples
examples = X_test_transformed.iloc[:10]

# Create a SHAP explainer for the Random Forest
explainer = shap.Explainer(
    rf_model,
    background,
    algorithm="tree"
)

shap_values = explainer(examples)

print("SHAP explanations generated successfully!")
print("Explanation shape:", shap_values.values.shape)
```
