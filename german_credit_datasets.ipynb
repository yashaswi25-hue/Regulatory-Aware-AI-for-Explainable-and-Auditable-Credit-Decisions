```python
target_text = (
    df["Credit Risk"]
    .astype(str)
    .str.strip()
    .str.lower()
)

target_mapping = {
    "1": 0,       # Good risk in original UCI coding
    "2": 1,       # Bad risk in original UCI coding
    "good": 0,
    "bad": 1
}

y = target_text.map(target_mapping)

if y.isna().any():
    raise ValueError(
        "Unexpected target labels. Check the original target values."
    )

X = df.drop(columns=["Credit Risk"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

numeric_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight=None
    ))
])

print("Preprocessing and model pipeline created.")
```

```python
model.fit(X_train, y_train)

print("Model training complete!")
```

```python
y_pred = model.predict(X_test)

print("Actual labels:")
print(y_test.head(10).tolist())

print("\nPredicted labels:")
print(y_pred[:10].tolist())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred, zero_division=0
)

recall = recall_score(
    y_test, y_pred, zero_division=0
)

f1 = f1_score(
    y_test, y_pred, zero_division=0
)

print(f"Accuracy:  {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall:    {recall * 100:.2f}%")
print(f"F1-score:  {f1 * 100:.2f}%")

print("\nFull classification report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Good credit risk", "Bad credit risk"],
    zero_division=0
))
```

```python


from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
import pandas as pd

# Preprocessing stays inside each cross-validation fold
numeric_cols = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = X_train.select_dtypes(exclude=["number"]).columns.tolist()

def make_preprocessor(scale_numeric=False):
    numeric_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale_numeric:
        numeric_steps.append(("scaler", StandardScaler()))

    numeric_pipe = Pipeline(numeric_steps)

    categorical_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipe, numeric_cols),
        ("categorical", categorical_pipe, categorical_cols)
    ])

models = {
    "Random Forest": (
        RandomForestClassifier(n_estimators=200, random_state=42),
        False
    ),
    "Decision Tree": (DecisionTreeClassifier(random_state=42), False),
    "Logistic Regression": (LogisticRegression(max_iter=3000), True),
    "SVM": (SVC(), True)
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "bad_credit_recall": "recall",
    "f1": "f1"
}

cv_results = []

for name, (classifier, scale_numeric) in models.items():
    pipeline = Pipeline([
        ("preprocessing", make_preprocessor(scale_numeric)),
        ("classifier", classifier)
    ])

    scores = cross_validate(
        pipeline, X_train, y_train,
        cv=cv, scoring=scoring, n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy (mean)": scores["test_accuracy"].mean(),
        "Balanced Accuracy (mean)": scores["test_balanced_accuracy"].mean(),
        "Bad-credit Recall (mean)": scores["test_bad_credit_recall"].mean(),
        "F1 (mean)": scores["test_f1"].mean(),
        "F1 (std)": scores["test_f1"].std()
    })

cv_results_df = pd.DataFrame(cv_results).sort_values(
    "F1 (mean)", ascending=False
)

display(cv_results_df.round(3))
cv_results_df.to_csv("stratified_5fold_model_comparison.csv", index=False)
print("5-fold cross-validation results saved.")
```

```python

%pip install -q shap

import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# Fit both models using training data only
lr_model = Pipeline([
    ("preprocessing", make_preprocessor(scale_numeric=True)),
    ("classifier", LogisticRegression(max_iter=3000))
])

rf_model = Pipeline([
    ("preprocessing", make_preprocessor(scale_numeric=False)),
    ("classifier", RandomForestClassifier(
        n_estimators=200, random_state=42
    ))
])

lr_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)

# Transformed feature names
feature_names = lr_model.named_steps[
    "preprocessing"
].get_feature_names_out()

# Logistic Regression coefficients
lr_coefficients = lr_model.named_steps["classifier"].coef_[0]

lr_importance = pd.DataFrame({
    "Feature": feature_names,
    "LR Coefficient": lr_coefficients,
    "LR Absolute Importance": np.abs(lr_coefficients)
}).sort_values("LR Absolute Importance", ascending=False)

print("TOP 15 LOGISTIC REGRESSION FEATURES")
display(lr_importance.head(15).round(4))

# Random Forest SHAP explanations
X_train_rf = rf_model.named_steps["preprocessing"].transform(X_train)
X_test_rf = rf_model.named_steps["preprocessing"].transform(X_test)

explainer = shap.TreeExplainer(
    rf_model.named_steps["classifier"]
)
shap_values = explainer.shap_values(X_test_rf)

# Handle different SHAP output formats
if isinstance(shap_values, list):
    bad_credit_shap = shap_values[1]
elif np.asarray(shap_values).ndim == 3:
    bad_credit_shap = np.asarray(shap_values)[:, :, 1]
else:
    bad_credit_shap = np.asarray(shap_values)

rf_importance = pd.DataFrame({
    "Feature": feature_names,
    "RF Mean Absolute SHAP": np.abs(bad_credit_shap).mean(axis=0)
}).sort_values("RF Mean Absolute SHAP", ascending=False)

print("TOP 15 RANDOM FOREST SHAP FEATURES")
display(rf_importance.head(15).round(4))

# Compare top features
comparison = lr_importance.merge(
    rf_importance, on="Feature", how="outer"
).fillna(0)

comparison["LR Rank"] = comparison[
    "LR Absolute Importance"
].rank(ascending=False, method="min")

comparison["RF SHAP Rank"] = comparison[
    "RF Mean Absolute SHAP"
].rank(ascending=False, method="min")

print("FEATURE IMPORTANCE COMPARISON")
display(
    comparison.sort_values("LR Absolute Importance", ascending=False)
    [["Feature", "LR Coefficient", "LR Absolute Importance",
      "RF Mean Absolute SHAP", "LR Rank", "RF SHAP Rank"]]
    .head(15).round(4)
)

# Visualize the top 10 features from each explanation
lr_importance.head(10).sort_values(
    "LR Absolute Importance"
).plot(
    x="Feature", y="LR Absolute Importance",
    kind="barh", legend=False, figsize=(8, 5)
)
plt.title("Logistic Regression: Absolute Coefficients")
plt.tight_layout()
plt.show()

rf_importance.head(10).sort_values(
    "RF Mean Absolute SHAP"
).plot(
    x="Feature", y="RF Mean Absolute SHAP",
    kind="barh", legend=False, figsize=(8, 5)
)
plt.title("Random Forest: Mean Absolute SHAP")
plt.tight_layout()
plt.show()
```

```python

from datetime import datetime, timezone
import shap

MODEL_VERSION = "RF-1.0"
MODEL_NAME = "Random Forest"
AUDIT_DATE = datetime.now(timezone.utc).isoformat()

# Transform test data using the pipeline's fitted preprocessing
X_transformed = rf_model[:-1].transform(X_test)
feature_names = rf_model[:-1].get_feature_names_out()

# Explain the Random Forest classifier
rf_classifier = rf_model.steps[-1][1]
explainer = shap.TreeExplainer(rf_classifier)
shap_values = explainer.shap_values(X_transformed)

# Support common SHAP output formats
if isinstance(shap_values, list):
    contributions = np.asarray(shap_values[1])
else:
    contributions = np.asarray(shap_values)
    if contributions.ndim == 3:
        contributions = contributions[:, :, 1]

predictions = rf_model.predict(X_test)
probabilities = rf_model.predict_proba(X_test)[:, 1]

log_rows = []

for i in range(len(X_test)):
    top_indices = np.argsort(np.abs(contributions[i]))[-3:][::-1]

    explanation_parts = []
    for j in top_indices:
        direction = (
            "increases predicted bad-risk score"
            if contributions[i, j] > 0
            else "decreases predicted bad-risk score"
        )
        explanation_parts.append(
            f"{feature_names[j]}: {direction}"
        )

    log_rows.append({
        "Applicant reference": f"TEST_{i + 1:04d}",
        "Model": MODEL_NAME,
        "Model version": MODEL_VERSION,
        "Audit timestamp UTC": AUDIT_DATE,
        "Predicted class": int(predictions[i]),
        "Predicted bad-risk probability": round(float(probabilities[i]), 4),
        "Top feature contributions": "; ".join(explanation_parts)
    })

decision_log = pd.DataFrame(log_rows)

display(decision_log.head(10))

decision_log.to_csv("credit_decision_audit_log.csv", index=False)
print("Saved: credit_decision_audit_log.csv")
```
